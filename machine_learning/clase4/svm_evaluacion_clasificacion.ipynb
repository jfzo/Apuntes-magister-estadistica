# Evaluación y selección de modelos

El título de esta parte de la sesión comprende dos objetivos distintos:

1. Evaluación de un modelo: Dado un método con sus parámetros ya definidos, estimar su capacidad de generalización usando datos nuevos.
2. Selección de un modelo: Estimar el desempeño de distintos métodos, usando las mismas condiciones experimentales, para escoger aquel con mejor desempeño.

## Medición del desempeño de un modelo

La calidad de las predicciones de un modelo sobre datos no vistos en la fase de entrenamiento (_construcción del modelo_) es extremadamente importante.

Poder medir este desempeño nos entrega una idea de que tan bien se comportará el modelo escogido en un escenario real.

__Contexto:__ Tenemos una variable dependiente (_target_) $Y$, un conjunto de observaciones de entrada $X$, y un modelo de predicción $\hat{f}(X)$ que ha sido estimado a partir de un conjunto de entrenamiento $\mathcal{T}$.

Contabilizaremos el error de este modelo, es decir desacuerdos entre $Y$ y $\hat{f}(X)$ mediante una denominada función de pérdida $L(Y, \hat{f}(X))$.

Algunas elecciones comunes para esta función de pérdida cuando $Y\in\mathbb{R}$son:

* $L(Y, \hat{f}(X)) = (Y - \hat{f}(X))^2$ : Error cuadrático
* $L(Y, \hat{f}(X)) = |Y - \hat{f}(X)|$ : Error absoluto

y cuando $Y$ es cualitativa o categorica:
* $L(Y, \hat{f}(X)) = \mathcal{I}(Y \neq \hat{f}(X))$

donde 
$ \mathcal{I}(a,b) =
  \begin{cases}
    1 & \text{if $a\neq b$} \\
    0 & \text{if $a=b$} 
  \end{cases}$
  
El __error de entrenamiento__ es el promedio de la función de pérdida sobre los datos de entrenamiento, es decir para $X=\{x_1,x_2,\ldots , x_N\}$ se tiene 
$$\overline{err}=\frac{1}{N}\sum_{i=1}^N L(y_i,\hat{f}(x_i))$$

El __error de prueba__ o de _test_ , también llamado error de generalización, corresponde al error de predicción sobre un conjunto de datos de prueba independiente.

El objetivo de esta sección es revisar al menos una manera de poder estimar el error de prueba, es decir $$\text{Err}_{\mathcal{T}}=E[L(Y, \hat{f}(X))|\mathcal{T}]$$
lo que equivale a el error de predicción cometido por el modelo entrenado sobre $\mathcal{T}$.

## Metricas de desempeño

* Matriz de confusión y derivadas

### Matriz de Confusión

La matriz que se observa se denomina **Matriz de Confusión**.

* Elementos en la diagonal representan a los puntos que fueron correctamente clasificados
* Elementos fuera de la diagonal representan a los puntos _incorrectamente_ clasificados

<img src="https://www.dropbox.com/s/ph693v8ee73nqmm/confusion_matrix.png?raw=1" width="500px"/>


Para la clasificación binaria, existen dos medidas muy usadas sobre todo en estudios relacionados con salud:

* Sensibilidada (_sensitivity_): porcentaje de individuos de la clase positiva correctamente identificados
* Especificidad (_specificity_): porcentaje de individuos de la clase negativa correctamente identificados


Asociadas a estas medidas también están:

* True-Positive rate: Equivale a la *sensibilidad*
* False-Positive rate: Equivale a $(1-$*especificidad*$)$

Otras medidas también usadas (no solamente para clasificación binaria) son:

* accuracy: Número de aciertos dividido por el total de ejemplos
* precision: Cantidad de aciertos dividido por total de ponósticos por clase (Qué proporción del total identificado como positivo fue correctamente identificado) $$P=\frac{Tp}{Tp+Fp}$$
* recall: Cantidad de aciertos dividido por total de ejemplos por clase (Qué proporción del total de positivos fueron identificados como tal)$$R=\frac{Tp}{Tp+Fn}$$
* F1: Media armónica entre precisión y recall $$2\cdot\frac{P\cdot R}{P+R}$$


Existen dos conceptos mencionados en el título de esta sección que corresponden a dos objetivos diferentes:

1. __Selección del modelo:__ Estimar el desempeño de varios modelos para poder escoger el mejor.
2. __Evaluación del modelo:__ Habiendo escogido el modelo final, estimar su error de predicción sobre datos nuevos.


## El enfoque del conjunto de validación

El conjunto de datos se divide aleatoriamente en 3 partes


El __subconjunto de entrenamiento__ permite generar el modelo.

El __subconjunto de validación__ permite escoger la combinación de parémtros del modelo que menor error alcanza.

El __subconjunto de prueba__ permite someter el modelo a datos nunca antes procesados y de esa manera medir su capacidad de generalizar.

**Importante** 

* En muchos casos reales, esta manera de dividir el conjunto de datos para realizar la selección y evaluación del modelo es impracticable debido a a la __baja cantidad__ de datos disponibles. Existen también otros mecanismos como por ejemplo la Validación Cruzada de $k$-folds.

* Distintas particiones del conjunto de datos entregaran resultados muy variables dependiendo de qué observaciones hayan quedado en cada segmento.
* Para estimar el error de prueba habrá ejemplos que no van a ser usados para entrenar el método

## Enfoque de k-Folds

Plantea dividir los datos en $k$ bloques, denominados *folds*, de aproximadamente el mismo tamaño.

* `IDEA INICIAL: `
El primer *fold* es usado para validación y los $(k-1)$ restantes, $\{2,3\ldots k\}$, para entrenar el método. Luego, se calcula el error promedio sobre los datos del *fold*-$1$, es decir $L_1$.

`El procedimiento anterior se repite` $k$ `veces, en cada una tomando un fold distinto para validación.`

Todo este proceso resultará en $k$ estimaciones del error de prueba, es decir $\{L_1,L_2\ldots , L_k\}$

Por lo tanto el estimado de este error generado por el procedimiento de $k$-fold *cross-validation* corresponderá al promedio de las estimaciones
$$CV_{(k)}=\frac{1}{k}\sum_{i=1}^k L_i$$

* **Caso especial** Cuando el valor de $k$ es igual a $n$ (cantidad de ejemplos), el procedimiento se conoce como *Leave-One-Out Cross-Validation*.
